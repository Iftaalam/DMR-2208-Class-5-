```python
#Import Pandas Package for data manipulation and analysis
import pandas as pd
```

```python
#Now load the dataset which want to work .
#Here dataset is a name of variable which we can set according to our will
dataset=pd.read_excel("Lightning.xlsx")
```

```python
#Check the dataset to ensure it is loaded or not by calling the assigned variable
dataset
```

```python
# We want to delete the unnecessary column . So let's see the heading of the column and decide what to delete
dataset.columns
```

```python
# List of columns to delete
columns_to_drop = [
    'start', 
    'end', 
    '_id', 
    '_uuid', 
    '_submission_time', 
    '_validation_status', 
    '_notes', 
    '_status', 
    '_submitted_by', 
    '__version__', 
    '_tags', 
    '_index',
    '(অনুগ্রহ করে নিচের প্রতিটি বিবৃতি পড়ুন এবং উত্তর গুলোর মধ্যে গত সপ্তাহ ব্যাপী আপনার জন্য যা প্রযোজ্য তা নির্বাচন করুন। এখানে কোন সঠিক বা ভুল উত্তর নেই। কোন বিবৃতির জন্য বেশী সময় ব্যয় করবেন না। ধন্যবাদ)',
    'Translated and Validated by Dr S M Abu Hena Mostafa Alim, BSMMU, Bangladesh.'
]

# Drop the columns and save the result back to 'dataset'
dataset = dataset.drop(columns=columns_to_drop)

# Check the dataset again to ensure they are gone
dataset.head()
```

```python
#Code Explanation
##\u0900-\u09FF]+: This is the universal Unicode range for Bengai characters and punctuation. The code finds anything in this range and deletes it.

##\([\s-]*\): When a cell gets stripped of Bangla, it turns into (-). This line finds those leftover brackets and hyphens and deletes them so you are just left with 36-45.

##strip(): This ensures that any awkward spaces left at the end of words (like "Male " instead of "Female") are removed.
    
    
# 1. Remove all Bangla characters from the Column Names
dataset.columns = dataset.columns.str.replace(r'[\u0900-\u09FF]+', '', regex=True)
dataset.columns = dataset.columns.str.replace(r'\([\s-]*\)', '', regex=True) # Cleans up empty ()
dataset.columns = dataset.columns.str.replace(r'\[[\s-]*\]', '', regex=True) # Cleans up empty []
dataset.columns = dataset.columns.str.strip() # Removes extra spaces

# 2. Remove all Bangla characters from the Data Rows
dataset = dataset.replace(r'[\u0900-\u09FF]+', '', regex=True)
dataset = dataset.replace(r'\([\s-]*\)', '', regex=True)
dataset = dataset.replace(r'\[[\s-]*\]', '', regex=True)

# 3. Strip extra spaces left behind in the text cells
dataset = dataset.applymap(lambda x: x.strip() if isinstance(x, str) else x)

# Check the cleaned dataset
dataset.head()
```

```python
# Now we will rename the column , there are lots of way to do that
#First , you can use the column name directly
dataset = dataset.rename(columns={'Age': 'Current_Age'})
dataset
```

```python
# Second ,you can rename the column by picking the column position
# Provide a list matching the exact number of columns you currently have
dataset.columns = ['Gender', 'Age', 'Education', 'Income', 'K1', 'K2', 'A1', 'A2', 'P1', 'P2']
dataset
```

```python
# Remove the existing parentheses that contain spaces, hyphens, OR commas
dataset = dataset.replace(r'\([\s,-]*\)', '', regex=True)

# Run the strip command again to clean up the extra space left at the end
dataset = dataset.applymap(lambda x: x.strip() if isinstance(x, str) else x)

# Check the dataset
dataset.head()
```

```python
dataset
```

```python
# Loop through every single column in the dataset
for column in dataset.columns:
    
    # 1. Get the raw count of each category
    counts = dataset[column].value_counts()
    
    # 2. Get the percentage (normalize=True gives decimals like 0.71, so we multiply by 100)
    percentages = dataset[column].value_counts(normalize=True) * 100
    
    # 3. Round the percentage to 2 decimal places so it looks clean
    percentages = percentages.round(2)
    
    # 4. Combine them into the format: "Count (Percentage%)"
    combined_result = counts.astype(str) + " (" + percentages.astype(str) + "%)"
    
    # Print the column name as a header, then the results
    print(f"--- {column} ---")
    print(combined_result)
    print("\n") # This just adds a blank line between columns for readability
```

```python
# Create the dictionary and name it 'my_scores'
Quantification= {
    'Agree': 3,
    'Neutral': 2,
    'Disagree': 1
}

# Use the exact same name 'Quantification' inside the replace function
dataset['K1'] = dataset['K1'].replace(Quantification)
dataset
```

```python
# Assign value for other column
Assigning_Value = ['K2', 'A1', 'A2','P1','P2']
dataset[Assigning_Value] = dataset[Assigning_Value].replace(Quantification)
```

```python
dataset
```

```python
# 1. Calculate Knowledge Score (Sum of K1 and K2)
dataset['Knowledge_Score'] = dataset['K1'] + dataset['K2']

# 2. Calculate Attitude Score (Sum of A1 and A2)
dataset['Attitude_Score'] = dataset['A1'] + dataset['A2']

# 3. Calculate Practice Score (Sum of P1 and P2)
dataset['Practice_Score'] = dataset['P1'] + dataset['P2']

# 4. View the first few rows of your new scores to verify!
dataset[['Knowledge_Score', 'Attitude_Score', 'Practice_Score']].head()
```

```python
dataset
```

```python
# Select your score columns and ask for the 'mean' and 'std' (standard deviation)
score_stats = dataset[['Knowledge_Score', 'Attitude_Score', 'Practice_Score']].agg(['mean', 'std'])

# Round the numbers to 2 decimal places so they are easy to read
score_stats = score_stats.round(2)

# Print the results
print("--- Mean and Standard Deviation ---")
print(score_stats)
```

```python
# Group by Gender, then calculate the mean and std for the score columns
gender_stats = dataset.groupby('Gender')[['Knowledge_Score', 'Attitude_Score', 'Practice_Score']].agg(['mean', 'std'])

# Round the numbers to 2 decimal places to keep it neat
gender_stats = gender_stats.round(2)

# Print the results
print("--- Mean and Standard Deviation by Gender ---")
print(gender_stats)
```

```python
# Calculate the Pearson correlation
correlation = dataset['Knowledge_Score'].corr(dataset['Attitude_Score'], method='pearson')

# Print the result, rounding to 3 decimal places
print(f"Pearson Correlation (Knowledge vs Attitude): {correlation:.3f}")
```

```python
# Generate a correlation matrix for all three scores
correlation_matrix = dataset[['Knowledge_Score', 'Attitude_Score', 'Practice_Score']].corr(method='pearson')

# Print the rounded matrix
print("--- Pearson Correlation Matrix ---")
print(correlation_matrix.round(3))
```

```python
# Calculate the Spearman rank correlation between Attitude and Practice
spearman_corr = dataset['Attitude_Score'].corr(dataset['Practice_Score'], method='spearman')

# Print the result, rounding to 3 decimal places
print(f"Spearman Correlation (Attitude vs Practice): {spearman_corr:.3f}")
```

```python
# Generate a Spearman correlation matrix for all three scores
spearman_matrix = dataset[['Knowledge_Score', 'Attitude_Score', 'Practice_Score']].corr(method='spearman')

# Round the results to 3 decimal places for readability
spearman_matrix = spearman_matrix.round(3)

# Print the matrix
print("--- Spearman Correlation Matrix ---")
print(spearman_matrix)
```

```python
# Import the stats module from the scipy library
from scipy import stats

# Run the simple linear regression
# We put the "Predictor" (Knowledge) first, and the "Outcome" (Practice) second
result = stats.linregress(dataset['Knowledge_Score'], dataset['Practice_Score'])

# Print the full results beautifully
print("--- Simple Linear Regression Results ---")
print(f"Slope (Coefficient): {result.slope:.3f}")
print(f"Intercept: {result.intercept:.3f}")
print(f"R-squared: {(result.rvalue**2):.3f}")
print(f"P-value: {result.pvalue:.4f}")
```

```python
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf

# 1. Define models
f1 = "Knowledge_Score ~ C(Gender, Treatment('Male')) + C(Age, Treatment('18-25')) + C(Education, Treatment('Illiterate')) + C(Income, Treatment('No income'))"
f2 = "Attitude_Score ~ C(Gender, Treatment('Male')) + C(Age, Treatment('18-25')) + C(Education, Treatment('Illiterate')) + C(Income, Treatment('No income'))"
f3 = "Practice_Score ~ C(Gender, Treatment('Male')) + C(Age, Treatment('18-25')) + C(Education, Treatment('Illiterate')) + C(Income, Treatment('No income'))"

m1 = smf.ols(formula=f1, data=dataset).fit()
m2 = smf.ols(formula=f2, data=dataset).fit()
m3 = smf.ols(formula=f3, data=dataset).fit()

# 2. Helper function to format coefficients with stars and NO standard errors
def format_coef(val, pval):
    if pd.isna(val): return ""
    stars = ""
    if pval < 0.01: stars = "***"
    elif pval < 0.05: stars = "**"
    elif pval < 0.1: stars = "*"
    return f"{val:.2f}{stars}"

# 3. Create the exact row structure you want manually
rows = [
    ("1. Gender", ""),
    ("    Male", "REF"),
    ("    Female", "C(Gender, Treatment('Male'))[T.Female]"),
    ("2. Age group (year)", ""),
    ("    18-25", "REF"),
    ("    26-35", "C(Age, Treatment('18-25'))[T.26-35]"),
    ("    36-45", "C(Age, Treatment('18-25'))[T.36-45]"),
    ("    46-55", "C(Age, Treatment('18-25'))[T.46-55]"),
    ("    Less than 18", "C(Age, Treatment('18-25'))[T.Less than 18]"),
    ("    More than 55", "C(Age, Treatment('18-25'))[T.more than 55]"),
    ("3. Education", ""),
    ("    Illiterate", "REF"),
    ("    Non-SSC", "C(Education, Treatment('Illiterate'))[T.Non-SSC]"),
    ("    SSC", "C(Education, Treatment('Illiterate'))[T.SSC]"),
    ("    HSC", "C(Education, Treatment('Illiterate'))[T.HSC]"),
    ("    Bachelor/Undergraduate", "C(Education, Treatment('Illiterate'))[T.BachelorUndergraduate]"),
    ("4. Monthly Income", ""),
    ("    No income", "REF"),
    ("    Less than 15000", "C(Income, Treatment('No income'))[T.Less than 15000]"),
    ("    15000-29999", "C(Income, Treatment('No income'))[T.15000-29999]"),
    ("    30000-49999", "C(Income, Treatment('No income'))[T.30000-49999]")
]

# 4. Extract data and build table
data = []
for label, key in rows:
    if key == "" or key == "REF":
        data.append([label, "", "", ""])
    else:
        v1, p1 = m1.params.get(key, np.nan), m1.pvalues.get(key, 1.0)
        v2, p2 = m2.params.get(key, np.nan), m2.pvalues.get(key, 1.0)
        v3, p3 = m3.params.get(key, np.nan), m3.pvalues.get(key, 1.0)
        data.append([label, format_coef(v1, p1), format_coef(v2, p2), format_coef(v3, p3)])

# Print as DataFrame
out_df = pd.DataFrame(data, columns=['Features', 'Model 1 Knowledge', 'Model 2 Attitude', 'Model 3 Practice'])
print(out_df.to_string(index=False))
```
